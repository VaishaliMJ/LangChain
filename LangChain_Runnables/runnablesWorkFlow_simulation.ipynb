```python
import random
class sampleLLM:
    def __init__(self):
        print("LLM Created")
    
    def predict(self,prompt):
        responseList=[
            "Python is a Programming Language",
            "GenAI stands for Generative AI",
            "AI Stands for Artifical Intelligence"
        ]
        
        return {"response":random.choice(responseList)}
    
     
```

```python
llm=sampleLLM()
```

```python
llm.predict("What is AI?")
```

```python
class samplePromptTemplate:
    def __init__(self,template,inputVariables):
        self.template=template
        self.inputVariables=inputVariables
        
    def format(self,inputDict):
        return self.template.format(**inputDict)
    
        
```

```python
template=samplePromptTemplate(
    template="Write a {length} poem on topic {topic}",
    inputVariables=["length","topic"]
)
```

```python
prompt=template.format({"topic":"India","length":"short"})
```

```python
llmModel=sampleLLM()
```

```python
result=llmModel.predict(prompt)

print(result)
```

```python
class sampleLLMChain:
    def __init__(self,llm,prompt):
        self.llm=llm
        self.prompt=prompt
        
    def run(self,inputDict):
        finalPrompt=self.prompt.format(inputDict) 
        llmResult=self.llm.predict(finalPrompt)
        
        return llmResult["response"]   
```

```python
# Sample testing of chain 
template=samplePromptTemplate(
    template="Write a {length} poem on topic {topic}",
    inputVariables=["length","topic"]
)

llmModel=sampleLLM()


chain=sampleLLMChain(llmModel,template)

chain.run({'length':'Long', 'topic':'Footbal'})
```
