```python
from abc import ABC,abstractmethod
```

```python
class sampleRunnable(ABC):
    @abstractmethod
    def invoke(input_data):
        pass
```

```python
import random
class sampleLLM(sampleRunnable):
    def __init__(self):
        print("LLM Created")
    
    
    def invoke(self,prompt):
        responseList=[
            "Python is a Programming Language",
            "GenAI stands for Generative AI",
            "AI Stands for Artifical Intelligence"
        ]
        
        return {"response":random.choice(responseList)}
    
    
    def predict(self,prompt):
        responseList=[
            "Python is a Programming Language",
            "GenAI stands for Generative AI",
            "AI Stands for Artifical Intelligence"
        ]
        
        return {"response":random.choice(responseList)}
    
     
```

```python
class samplePromptTemplate(sampleRunnable):
    def __init__(self,template,inputVariables):
        self.template=template
        self.inputVariables=inputVariables
      
    def invoke(self,inputDict):
        return self.template.format(**inputDict)
        
    def format(self,inputDict):
        return self.template.format(**inputDict)
    
        
```

```python
template=samplePromptTemplate(
    template="Write a {length} poem on topic {topic}",
    inputVariables=["length","topic"]
)
```

```python
prompt=template.format({"topic":"India","length":"short"})
```

```python
llmModel=sampleLLM()
```

```python
result=llmModel.predict(prompt)

print(result)
```

```python
class sampleStrOutputParser(sampleRunnable):
    def __init__(self):
        pass  
      
    def invoke(self,inputData):
        return inputData["response"]  
```

```python
class sampleRunnableConnector(sampleRunnable):
    def __init__(self,runnableList):
        self.runnableList=runnableList
    
    
    def invoke(self,inputData):
        for runnable in self.runnableList:
            inputData=runnable.invoke(inputData) 
        return inputData       
```

```python
testTemplate_1=samplePromptTemplate(
    template="Write a poem on topic {topic}",
    inputVariables=["topic"]
)

testTemplate_2=samplePromptTemplate(
    template="Write a summary of  the following  {response}",
    inputVariables=["response"]
)

testLLMModel=sampleLLM()

```

```python
testParser = sampleStrOutputParser()

chain_1=sampleRunnableConnector([testTemplate_1,testLLMModel])

chain_2=sampleRunnableConnector([testTemplate_2,testLLMModel,testParser])


```

```python
finalChain=sampleRunnableConnector([chain_1,chain_2])


result=finalChain.invoke({'topic':'cricket'})
print(result)
```
